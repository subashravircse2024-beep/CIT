```python
# ============================================
# BASIC RAG USING SENTENCE TRANSFORMERS + FAISS
# + FLAN-T5
# ============================================

from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
from transformers import pipeline

# --------------------------------------------
# 1. KNOWLEDGE BASE
# --------------------------------------------

documents = [
    "The Eiffel Tower is located in Paris, France and was completed in 1889.",

    "Retrieval-Augmented Generation combines document retrieval with text generation.",

    "Python is a popular high-level programming language used in AI development.",

    "Vector databases store embeddings and support fast similarity search."
]

print("Number of documents:", len(documents))


# --------------------------------------------
# 2. LOAD EMBEDDING MODEL
# --------------------------------------------

print("\nLoading embedding model...")

embed_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully.")


# --------------------------------------------
# 3. CREATE DOCUMENT EMBEDDINGS
# --------------------------------------------

print("\nCreating document embeddings...")

doc_embeddings = embed_model.encode(
    documents,
    convert_to_numpy=True
)

print("Embedding shape:", doc_embeddings.shape)


# --------------------------------------------
# 4. BUILD FAISS INDEX
# --------------------------------------------

dimension = doc_embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(doc_embeddings)

print("\nFAISS index created.")
print("Number of vectors in index:", index.ntotal)


# --------------------------------------------
# 5. USER QUERY
# --------------------------------------------

query = "What is RAG in AI?"

print("\nUser Query:")
print(query)


# --------------------------------------------
# 6. EMBED USER QUERY
# --------------------------------------------

query_embedding = embed_model.encode(
    [query],
    convert_to_numpy=True
)


# --------------------------------------------
# 7. RETRIEVE TOP-K DOCUMENTS
# --------------------------------------------

k = 2

D, I = index.search(query_embedding, k)

retrieved_chunks = [
    documents[i]
    for i in I[0]
]

print("\nRetrieved Documents:")

for rank, (doc, distance) in enumerate(
    zip(retrieved_chunks, D[0]),
    start=1
):
    print(f"\nRank {rank}")
    print("Distance:", distance)
    print("Document:", doc)


# --------------------------------------------
# 8. CREATE CONTEXT
# --------------------------------------------

context = "\n".join(retrieved_chunks)

print("\nContext:")
print(context)


# --------------------------------------------
# 9. CREATE AUGMENTED PROMPT
# --------------------------------------------

prompt = f"""
Use the following context to answer the question.

Context:
{context}

Question:
{query}

Answer:
"""

print("\nAugmented Prompt:")
print(prompt)


# --------------------------------------------
# 10. LOAD GENERATIVE MODEL
# --------------------------------------------

print("Loading FLAN-T5 model...")

generator = pipeline(
    "text-generation",
    model="gpt2"
)

print("Generator loaded successfully.")


# --------------------------------------------
# 11. GENERATE FINAL ANSWER
# --------------------------------------------

result = generator(
    prompt,
    max_new_tokens=100,
    do_sample=False
)

answer = result[0]["generated_text"]


# --------------------------------------------
# 12. DISPLAY FINAL RESULT
# --------------------------------------------

print("\n============================================")
print("                RAG RESULT")
print("============================================")

print("\nQuery:")
print(query)

print("\nRetrieved Context:")
for chunk in retrieved_chunks:
    print("- ", chunk)

print("\nGenerated Answer:")
print(answer)

print("\n============================================")
```

```python
# ============================================================
# BASIC RAG SYSTEM
# Sentence Transformers + FAISS + FLAN-T5
# ============================================================

from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
from transformers import pipeline


# ============================================================
# 1. KNOWLEDGE BASE
# ============================================================

documents = [
    "The Eiffel Tower is located in Paris, France and was completed in 1889.",

    "Retrieval-Augmented Generation (RAG) is an AI technique that retrieves relevant information from a knowledge base and uses that information as context for a language model to generate an answer.",

    "Python is a popular high-level programming language used in AI development.",

    "Vector databases store numerical embeddings and support fast similarity search."
]

print("Number of documents:", len(documents))


# ============================================================
# 2. LOAD SENTENCE TRANSFORMER
# ============================================================

print("\nLoading embedding model...")

embed_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully.")


# ============================================================
# 3. CREATE DOCUMENT EMBEDDINGS
# ============================================================

print("\nCreating document embeddings...")

doc_embeddings = embed_model.encode(
    documents,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Embedding shape:", doc_embeddings.shape)


# ============================================================
# 4. BUILD FAISS VECTOR INDEX
# ============================================================

dimension = doc_embeddings.shape[1]

# IndexFlatIP + normalized embeddings
# gives cosine similarity search

index = faiss.IndexFlatIP(dimension)

index.add(
    doc_embeddings.astype("float32")
)

print("\nFAISS index created.")
print("Number of vectors in index:", index.ntotal)


# ============================================================
# 5. USER QUERY
# ============================================================

query = "What is RAG in AI?"

print("\nUser Query:")
print(query)


# ============================================================
# 6. CONVERT QUERY INTO EMBEDDING
# ============================================================

query_embedding = embed_model.encode(
    [query],
    convert_to_numpy=True,
    normalize_embeddings=True
)


# ============================================================
# 7. SEARCH FAISS FOR TOP-K DOCUMENTS
# ============================================================

k = 2

scores, indices = index.search(
    query_embedding.astype("float32"),
    k
)

retrieved_chunks = [
    documents[i]
    for i in indices[0]
]


# ============================================================
# 8. DISPLAY RETRIEVED DOCUMENTS
# ============================================================

print("\nRetrieved Documents:")

for rank, (doc, score) in enumerate(
    zip(retrieved_chunks, scores[0]),
    start=1
):

    print("\nRank", rank)
    print("Similarity:", round(float(score), 4))
    print("Document:", doc)


# ============================================================
# 9. CREATE CONTEXT
# ============================================================

context = "\n".join(
    retrieved_chunks
)

print("\nContext:")
print(context)


# ============================================================
# 10. CREATE AUGMENTED PROMPT
# ============================================================

prompt = f"""
Use the following context to answer the question.

Context:
{context}

Question:
{query}

Answer:
Use only information from the context.
"""


print("\nAugmented Prompt:")
print(prompt)


# ============================================================
# 11. LOAD GENERATIVE MODEL
# ============================================================

print("\nLoading generative model (gpt2)...")

generator = pipeline(
    "text-generation",
    model="gpt2"
)

print("Generator loaded successfully.")


# ============================================================
# 12. GENERATE ANSWER
# ============================================================

result = generator(
    prompt,
    max_new_tokens=100,
    do_sample=False
)

answer = result[0]["generated_text"]


# ============================================================
# 13. DISPLAY FINAL RAG RESULT
# ============================================================

print("\n")
print("============================================")
print("                RAG RESULT")
print("============================================")

print("\nQuery:")
print(query)

print("\nRetrieved Context:")

for chunk in retrieved_chunks:
    print("-", chunk)

print("\nGenerated Answer:")
print(answer)

print("\n============================================")
print("              END OF RAG")
print("============================================")
```
